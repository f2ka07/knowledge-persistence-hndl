# Threshold sensitivity

Stability of the influence-decay rule across threshold grid. Prefer stable regions over most significant p-values.

```python
import json
from pathlib import Path

import polars as pl

ROOT = Path('..').resolve()
for proc in (ROOT / 'dataset_v0.2_candidate' / 'processed', ROOT / 'data' / 'processed'):
  pq = proc / 'threshold_sensitivity.parquet'
  if pq.exists():
    GRID = pl.read_parquet(pq)
    break
else:
  raise FileNotFoundError('threshold_sensitivity.parquet not found')

for meta_base in (ROOT / 'dataset_v0.2_candidate' / 'metadata', ROOT / 'data' / 'metadata'):
  mj = meta_base / 'threshold_sensitivity.json'
  if mj.exists():
    stability = json.loads(mj.read_text(encoding='utf-8')).get('stability', {})
    break
else:
  stability = {}

display(GRID)
```

```python
region = GRID.filter(
  (pl.col('threshold_ratio').is_between(0.10, 0.15))
  & (pl.col('consecutive_years') == 3)
)
print('10-15% / 3-year region (tier-C log-rank p-values):')
print(region.select('threshold_ratio', 'consecutive_years', 'event_count', 'logrank_p_matched'))

print('\nStability recommendation:', stability.get('stable_region_recommendation'))
print('Tier C neutral at reference (10%, 3yr):', stability.get('tier_c_neutral_at_reference'))
```
